## 📚 Handling Null Values

```python
import pandas as pd
```

```python
df=pd.read_excel(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\NAData.xlsx")
```

```python
df
```

```python
df.isnull()
```

```python
# so true is 1 and false is 0. if we sum the true we will get the total number of null values in that column
# Null values is blank columns
df.isnull().sum()
```

```python
# 1. We can drop the rows having null values
df.dropna()
```

```python
# dropna method has argument thresh which means how many non-null values are allowed in a row.

df.dropna(thresh=5) # thresh=5 means it should have minimum 5 non-null values in a row if it's less it will drop that row
```

```python
# to run it columns-wise just try changing the axis

df.dropna(axis=1, thresh=8) # if there is less than 8 non-null values it will drop that column
```

```python
# we can replace the null value with a particular value

df.fillna(40) # this will replace all the null values with the assigned value
```

```python
df
```

```python
# we can also pass a dictionary to mention particular value for particular column

df.fillna({"Math": 40, "Hindi": 30})
```

```python
# so mean method generally gives the mean for that particular column
# use numeric_only=True to avoid string columns
df.mean(numeric_only=True)
```

```python
df.fillna(df.mean(numeric_only=True)) # this will replace null with mean of that column
```

```python
# so there are two separate methods which are used to handle null values
# first one is ffill and the other one is bfill

df.ffill()
```

```python
df.bfill()
```

```python
# to change it to row-wise instead of column change the axis
df.ffill(axis=1)
```

```python
# can keep the limit to repeat the same value to a particular limit

df.ffill(limit=2)
```

```python
df=df.iloc[:,1:]
```

```python
df
```

```python
# in pandas interpolate() is a method which fills the null values on the basis of the surrounding data points instead of a fixed constant value

df.interpolate()
```

### ✨ Replace
> There are two types of replace in pandas
1. Replacement at row-level. (Outer)
2. Replacement of string in a cell. (Inner)

```python
# suppose we have to simply replace Home Office to HO in segment

df["Segment"]=df["Segment"].replace("Home Office", "HO") # this will replace Home Office with HO and update the DataFrame
```

```python
df
```

```python
# Now wherever is 'o' in segment i want it to get it replaced by **.

df["Segment"]=df["Segment"].str.replace("o","**")
```

```python
df
```

```python
# now if we just want first 3 chars of the name
# we can slice it using the .str

df["F3_Chars"]=df["Customer Name"].str[:3]
```

```python
df
```

```python
# suppose we want last 3 characters of the Customer Name

df["L3"]=df["Customer Name"].str[-3:]
```

```python
df
```

```python
ndf=(df["Customer Name"].str.split(expand=True))
```

```python
ndf.columns=["First", "Last"]
```

```python
ndf
```

```python
df=df.join(ndf)
```
