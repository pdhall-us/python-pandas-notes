## 📚 Slicing of DataFrame
> 1. Slicing means extracting data subsets from the DataFrame. <br>
> 2. It has loc and iloc methods for index and label based slicing.

### ➡️**iloc method**
> Pandas iloc is a slicing method allows to extract specific set of rows and columns based on integer-based indexing.

```python
df.iloc[0] # Syntax: iloc[[rows_list], [columns_list]]
```

```python
df.iloc[[4]] # Row at index:4
```

```python
df.iloc[24:45] # extracting rows from rowID 25 to 45
```

```python
c=df.iloc[24,6] # access the value of cell at particular index [row_index, col_index]

print("Customer at Row-24: ",c)
```

```python
# slicing both rows and columns
df.iloc[10:21, :6] # 10 rows with 5 columns
```

### ➡️**loc method**
> Pandas loc is a method allows to extract specific set of rows and columns based on labels, slices or boolean conditions.

```python
df.loc[1:5] # loc method slicing doesn't have end-1, it gives directly from index 1 to 5
```

```python
df.loc[10:20,["Order ID", "Customer Name"]]
```

## 📚 Filtering
> Extracing specific rows based on some condtion.

```python
df.head()
```

```python
# now suppose we want order based on some specific category.
```

```python
u_val=df.Category.unique() # this will return unique values from Category column

for i,j in enumerate(u_val):
    print(f"{i}: {j}")
```

```python
df_tech=df[df.Category=='Technology']
df_tech
```

```python
df_tech[df_tech.Profit>0].head()
```

```python
# boolean operators for conditions more than one

df[(df["Category"]=="Technology") & df["Profit"]>0]
```

## 📚 Sorting
> Pandas df.sort_values() sort the values based on specified label. <br>
> df.sort_inde

```python
df.sort_values(by='Profit') # this will return records with profit in ascending order
```

```python
# if we want to sort in the original DataFrame with the highest to lowest profit
# then inplace=True will update the original DataFrame

df.sort_values(by='Profit', ascending=False, inplace=True)
df
```

```python
# if we want the original order

df.sort_index(inplace=True)
df
```
