## 📚 Data Visualization
> Data Visualization is the gaphical representation of data, so that it becomes easier to understand the trends and patterns in data.

### ✨ Matplotlib
> It is a visulization python library which is used for graphical representation of data.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
d={
    "name": ["Apple", "Orange", "Papaya", "Mango", "Grapes"],
    "price": [250,120,60,170,500]
}
```

```python
df=pd.DataFrame(d)

df
```

### ✨ What is Bar Chart?
> It is graphical representation of **Categorical** data using rectangular bars.<br>
> It is used to show comparison between the categories.

```python
plt.figure(figsize=(5,3)) # method to resize the visualization and figsize=(width, height)
plt.bar(df.name, df.price) 
plt.title("Name Vs Price", fontdict={"size":11, "weight": 400, "family": "serif"}, color="#f0520a")
plt.xlabel("Fruits", fontdict={"size":11, "weight": 400, "family": "serif"}, color="#f5424b")
plt.ylabel("Price", fontdict={"size":11, "weight": 400, "family": "serif"}, color="#f5424b")
plt.xticks(df.name, color="#4245f5", rotation=45)
plt.yticks([i for i in range(0,700,100)], color="#4245f5")
plt.grid(axis="y")
plt.show()
```

### ✨ What is Pie Chart?
> It is a graphical representation in circular format where each slice represents the proportion of the total data.

```python
plt.figure(figsize=(8,4))
plt.pie(df.price, labels=df.name, labeldistance=1.2, autopct="%1.1f%%", radius=0.8, colors=["#e31212","#fa7f14","#1ffa14","#faeb14","#aa14fa"])
plt.title("Name Vs Price",fontdict={"size":11,"weight":700,"family":"Serif"},color="#fa146c")
plt.show()
```

### ✨ What is scatter plot?
> It is a graphical representation which uses dots to show relationship between two numerical variables.<br>

#### ➡️ Key parts
1. **X-axis (Horizontal)**: Usually shows the independent variable (the cause or input).
2. **Y-axis (Vertical)**: Usually shows the dependent variable (the effect or output).
3. **Dots**: Each dot stands for a single data point or observation.

```python
plt.figure(figsize=(5,3))
plt.title("Age-Fare Survival Comparison", color="#fa1492", fontdict={"family": "sans-serif"})
plt.scatter(df.age, df.fare, color=["green" if i==1 else "red" for i in df.survived]) # green who survived and red who didn't
plt.xlabel("Age", fontdict={"size": 11, "weight": 400, "family": "sans-serif"}, color="#fa1414")
plt.ylabel("Fare", fontdict={"size": 11, "weight": 400, "family": "sans-serif"}, color="#fa1414")
plt.grid()
plt.show()
```

> As in the scatter plot a pattern which we can discover is people who bought expensive ticket are saved first as compared to the people who bought economy tickets.

```python
# Now let's dig deep into this and discover is there any male and female preference also while saving the passangers.
df1=df[df["sex"]=="male"]
df2=df[df["sex"]=="female"]
```

```python
plt.figure(figsize=(5,3))
plt.title("Age-Fare Survival Comparison", color="#fa1492", fontdict={"family": "sans-serif"})
plt.scatter(df1.age, df1.fare, color=["green" if i==1 else "red" for i in df1.survived]) # green who survived and red who didn't
plt.scatter(df2.age, df2.fare, color=["green" if i==1 else "red" for i in df2.survived], marker="x")
plt.xlabel("Age", fontdict={"size": 11, "weight": 400, "family": "sans-serif"}, color="#fa1414")
plt.ylabel("Fare", fontdict={"size": 11, "weight": 400, "family": "sans-serif"}, color="#fa1414")
plt.grid()
plt.show()
```

> So based on the scatter plot we can say that females were given preference over males while saving.

### ✨ What is Line Chart?
> It is a graphical representation in which data points are connected with line that shows the trend over a continuous progression or time.

#### ➡️ Key Components

1. **X-axis (Horizontal)**: Represents the sequential dimension, such as time (seconds, days, months, years) or ordered intervals.
2. **Y-axis (Vertical)**: Represents the quantitative metric or value being measured (e.g., temperature, sales, stock price).
3. **Data Points / Markers**: Individual values plotted at specific intersections on the grid.
4. **Line Segments**: Straight lines that join consecutive data points to reveal trajectories
