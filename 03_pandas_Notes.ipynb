```python
# now we have to add one more column of pass/fail in the original DataFrame

df=pd.read_excel(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\Data.xlsx")
```

```python
df
```

```python
df["Total"]=df.sum(axis=1, numeric_only=True)
```

```python
df
```

```python
df["Average"]=df["Total"]/5
```

```python
df
```

```python
# if the average is above 40 then the student is pass otherswise fail
# Method-1:

l=["Pass" if i>40 else "Fail" for i in df["Average"]]
```

```python
df["Pass/Fail"]=l
```

```python
df
```

```python
df.drop(columns="Pass/Fail", inplace=True)
```

```python
df
```

```python
# Method-2: by using apply method, it will automatically apply that function to each row at that column

def check(x):
    if x>40:
        return "Pass"
    else:
        return "Fail"
```

```python
df["Pass/Fail"]=df.Average.apply(check)
```

```python
df
```

```python
# now suppose student will be passed only if all the subjects marks are above 40. otherwise fail.

df.drop(columns="Pass/Fail", inplace=True)
```

```python
df
```

```python
def check1(x):
    if x["Math"]>=40 and x["Science"]>=40 and x["SST"]>=40 and x["Hindi"]>=40 and x["English"]>=40:
        return "Pass"
    else:
        return "Fail"
```

```python
df["Pass/Fail"]=df.apply(check1, axis=1)
```
