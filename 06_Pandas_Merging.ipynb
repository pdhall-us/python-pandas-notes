## 📚 Pandas Merging
> It is used to combine two DataFrames based on common keys using SQL-style join logic.<br>
> pd.merge()

```python
import pandas as pd
```

```python
personal=pd.read_csv(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\Personal.csv")
```

```python
professional=pd.read_csv(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\Professional.csv")
```

```python
personal
```

```python
professional
```

```python
# Now we have to join these two DataFrames based on the column Name

pd.merge(personal, professional, how="inner", left_on="Name", right_on="Name")
```

```python
# Left Join

pd.merge(personal, professional, how="left", left_on="Name", right_on="Name")
```

```python
left_anti=pd.merge(personal, professional, how="left", left_on="Name", right_on="Name", indicator=True)
```

```python
left_anti[left_anti["_merge"]=="left_only"]
```

```python
# Full outer join is also known as the universal join

pd.merge(personal, professional, how="outer", left_on="Name", right_on="Name", indicator=True)
```

```python
# from full outer join we can fetch all the other joins
```
