## 📚 Data Transformation
> It is the process of modifying and restructuring the data to prepare it for analysis. <br>
> It can also defined as changing the shape of DataFrame by using data transformation methods.

💡 Note:-
1. Slicing is not considered as data tranformation.
2. Just like in group by we are not changing the original DataFrame but do some tranformations, it is considered as Data Tranformation.

### 📚 Pivot Table

> Primarily pivot tables are used to summarize the large dataset into two-dimensional cleaner format.

```python
x=df.groupby(["Category", "Sub-Category", "Region"]).agg({"Sales": "sum"})
x.reset_index(inplace=True)
x
```

```python
# now suppose we have to create pivot for the above mentioned df
# for this we can use pivot_table() method, which is the pandas method

x=pd.pivot_table(data=df, index=["Category", "Sub-Category"], columns="Region", values="Sales", aggfunc="sum")
```

```python
x.reset_index(inplace=True)
```

```python
x
```

```python
x.columns.name=None
```

```python
x
```

```python
x.T # to transpose simply use capital T or transpose method
```

```python
x.transpose()
```

```python
x=df.groupby(["Category", "Sub-Category"]).agg({"Sales": "sum"})
x.reset_index(inplace=True)
```

```python
x
```

```python
# similar to lead() function in sql and if we need to calculate YoY or MoM, for this we can use shift(n) method

x["diff"]=x["Sales"]-x["Sales"].shift(1)
```

### 📚 Data Consolidation
> It is the process of combining two or more DataFrames or Series to make it unified DataFrame.

```python
df1=pd.read_excel(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\Consolidation\Jan-2024.xlsx")
```

```python
df1["Months"]="Jan"
```

```python
df1
```

```python
df2=pd.read_excel(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\Consolidation\Feb-2024.xlsx")
df2["Months"]="Feb"
```

```python
df2
```

```python
# concat() is the pandas method for combining two or more Dataframes row-wise or column-wise

df=pd.concat([df1,df2], ignore_index=True)
```

```python
path=r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\Consolidation"
file_list=os.listdir(path)
df=pd.DataFrame()
for i in file_list:
    final_path=os.path.join(path,i)
    df1=pd.read_excel(final_path)
    df1["Month"]=i.strip(".xlsx")
    df=pd.concat([df,df1], ignore_index=True)
```
