## 📚 Seaborn
> Seaborn is a python visualization library which is built on top of matplotlib that helps to create clean and attractive statistical data visualizations with minimal code.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import seaborn.objects as so
```

### ✨ Countplot
> It displays the frequency of observations in each categorical bins using bars.

```python
df=pd.read_csv(r"C:\Users\Prateek\Desktop\Repos\python-pandas-notes\sample_data\titanic passenger list.csv")
```

```python
df.head()
```

```python
plt.figure(figsize=(5,3))
sns.countplot(x=df.survived, hue=df.sex)
plt.xticks([0,1], ["Died", "Survived"])
plt.show()
```
