### ➡️ Sum for every 5th day, not the rolling sum
> so it will calculate the sum of values current date with the next 4days in that row and reduce it to that date.

```python
# step-1: to change the index to date

df=df.loc[:,["Ship Date", "Profit"]]
```

```python
df.head()
```

```python
df=df.set_index("Ship Date")
```

```python
# to calculate sum for every 5 days we use resample method

df["Profit"].resample("5D").sum().reset_index() # here it is giving sum on every 5th day
```

```python
df["Profit"].resample("YE").sum().reset_index() # here it is giving yearly profit
```

### ➡️ Rolling average of every 3rd day, which means current and previous 2
> for this we have to use rolling(window=3) with the mean method.

```python
df.reset_index(inplace=True)
```

```python
df.head()
```

```python
df["Rolling Profit"]=df["Profit"].rolling(window=3).mean()
```
