### ✨ Concatenate

```python
arr1=np.array([1,2,3,4])
arr2=np.array([5,6,7,8])
np.concatenate((arr1,arr2))
```

```python
arr1=np.array([[2,3,4],[10,20,30]])
arr2=np.array([[5,6,7],[50,60,70]])
arr3=np.concatenate((arr1,arr2)) # concatenating row-wise
arr3
```

```python
arr3.shape
```

```python
arr1=np.array([[2,3,4],[10,20,30]])
arr2=np.array([[2,6,7],[50,60,70]])
arr3=np.concatenate((arr1,arr2), axis=1) # concatenating column-wise
arr3
```

```python
arr3.shape
```

### ✨ Split

```python
l=[22,34,56,23,88,22,45,52,10,20,89,77]

l1=np.array(l)
np.array_split(l1,6)
```

### ✨ Where

```python
l2=l1.reshape(3,4)
np.where(l2==89) # this will provide with the index position of that number
```

```python
np.where(l2%2==0) # return all the indexes where the number is even
```

### ✨  Sorting

```python
l2
```

```python
np.sort(l2, axis=0) # vertically column-wise
```

```python
np.sort(l2, axis=1) # horizontally row-wise
```

```python
l=[22,34,56,23,88,22,45,52,10,20,89,77]

l1=np.array(l)
x=[]

for i in l1:
    if i%2==0:
        x.append(True)
    else:
        x.append(False)
```

```python
l1[x] # this is numpy syntax to return all the even numbers
```
